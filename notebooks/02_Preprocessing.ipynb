# 02 - Data Preprocessing

This notebook handles data cleaning, filtering, and dataset splitting.

```python
import sys
from pathlib import Path

# Add the project root to Python path for proper imports
project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))
sys.path.insert(0, str(project_root / 'src'))

import pandas as pd

from src.utils.config import *
from src.utils.helpers import set_random_seed
from src.preprocessing import clean_dataset, split_dataset

set_random_seed()
print("Preprocessing notebook ready!")
```

```python
# Load metadata
metadata_df = pd.read_csv(METADATA_FILE)
print(f"Loaded {len(metadata_df)} files")
```

```python
# Clean dataset
cleaned_df = clean_dataset(
    metadata_path=METADATA_FILE,
    min_notes=MIN_NOTES,
    min_duration=1.0,
    max_duration=600.0,
    balance=True,
    balance_method='undersample',
    output_path=INTERIM_DATA_DIR / 'cleaned_metadata.csv'
)
print(f"Cleaned dataset: {len(cleaned_df)} files")
```

```python
# Split dataset
train_df, val_df, test_df = split_dataset(
    cleaned_df,
    train_split=TRAIN_SPLIT,
    val_split=VAL_SPLIT,
    test_split=TEST_SPLIT,
    stratified=True,
    output_dir=INTERIM_DATA_DIR
)
print(f"Train: {len(train_df)}, Val: {len(val_df)}, Test: {len(test_df)}")
```
