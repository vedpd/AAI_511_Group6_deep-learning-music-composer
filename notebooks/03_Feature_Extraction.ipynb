# 03 - Feature Extraction

This notebook extracts features from MIDI files for both LSTM and CNN models.

```python
import sys
from pathlib import Path

# Add the project root to Python path for proper imports
project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

import pandas as pd
import numpy as np

from src.utils.config import *
from src.utils.config import ensure_directories
from src.utils.helpers import set_random_seed
from src.datasets import prepare_lstm_dataset, prepare_cnn_dataset

# Create output directories
ensure_directories()

set_random_seed()
print("Feature extraction notebook ready!")
```

```python
# Load split data
train_df = pd.read_csv(INTERIM_DATA_DIR / 'train_metadata.csv')
val_df = pd.read_csv(INTERIM_DATA_DIR / 'val_metadata.csv')
test_df = pd.read_csv(INTERIM_DATA_DIR / 'test_metadata.csv')
print(f"Loaded splits: Train {len(train_df)}, Val {len(val_df)}, Test {len(test_df)}")
```

```python
# Extract LSTM features
X_train_lstm, y_train_lstm, X_val_lstm, y_val_lstm, X_test_lstm, y_test_lstm = prepare_lstm_dataset(
    train_df, val_df, test_df,
    feature_type='combined',
    sequence_length=SEQUENCE_LENGTH,
    normalize=True,
    save_path=LSTM_FEATURES_FILE
)
print(f"LSTM features shape: Train {X_train_lstm.shape}, Val {X_val_lstm.shape}, Test {X_test_lstm.shape}")
```

```python
# Extract CNN features
X_train_cnn, y_train_cnn, X_val_cnn, y_val_cnn, X_test_cnn, y_test_cnn = prepare_cnn_dataset(
    train_df, val_df, test_df,
    roll_type='velocity',
    time_steps=MAX_TIME_STEPS,
    pitch_range=PITCH_RANGE,
    fps=20.0,
    save_path=CNN_FEATURES_FILE
)
print(f"CNN features shape: Train {X_train_cnn.shape}, Val {X_val_cnn.shape}, Test {X_test_cnn.shape}")
```
